```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style("whitegrid")

# Load Dataset
heart_df = pd.read_csv('heart.csv')
heart_df.head()
```

```python
plt.plot(heart_df["age"], heart_df["sex"])
plt.show()
```

```python
# 1. POINT_PLOT
sns.pointplot(x='cp', y='age', data=heart_df)
plt.show()
```

```python
# 2. REGRESSION_PLOT
sns.regplot(x='age', y='thalachh', data=heart_df)
plt.show()
```

```python
# 3. PAIR_PLOT
sns.pairplot(heart_df[["age","cp","thalachh"]],kind="hist", diag_kind="kde")
plt.suptitle('Pair Plot of Heart Disease Features')

plt.show()
```

```python
# 4. KDE_PLOT
sns.kdeplot(x=heart_df['oldpeak'], y=heart_df["age"], fill=True, color='r' )
plt.show()
```

```python
# 5. LINE_PLOT
plt.figure(figsize=(8, 5))
sns.lineplot(x='age', y='chol', data=heart_df,errorbar=None)
plt.title('Line Plot: Age vs Cholesterol')
plt.show()
```

```python
# 6. STRIP_PLOT
sns.stripplot(x='sex', y='age', data=heart_df)
plt.title('Strip Plot: Age Distribution by Sex')
plt.show()
```

```python
# 7. VIOLIN_PLOT
sns.violinplot(x='cp', y='age', hue='sex' ,data=heart_df)
plt.show()
```

```python
# 8. BOX_PLOT
plt.figure(figsize=(8, 5))
sns.boxplot(x='sex', y='chol', data=heart_df)
plt.title('Box Plot: Cholesterol by Sex')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(heart_df.corr(),annot=True, cmap='coolwarm')
plt.title("se")
plt.show()
```

```python
# 9. HEAT_MAP
numeric_cols = heart_df.select_dtypes(include='number')
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Heat Map: Correlation Matrix')
plt.show()
```

```python
# 10. SCATTER_PLOT
plt.figure(figsize=(8, 5))
sns.scatterplot(x='age', y='chol', data=heart_df)
plt.title('Scatter Plot: Age vs Cholesterol')
plt.show()
```

```python
# 11. HISTOGRAM
plt.figure(figsize=(8, 5))
sns.histplot(x=heart_df['age'], bins=6, kde=True)
plt.title('Histogram: Age Distribution')
plt.show()
```

```python
# 12. BAR_PLOT
plt.figure(figsize=(8, 5))
sns.barplot(x='cp', y='chol', data=heart_df)
plt.title('Bar Plot: Average Cholesterol by Chest Pain Type')
plt.show()
```

```python
# 13. BUBBLE_GRAPH
plt.figure(figsize=(8, 5))
sns.scatterplot(x='age', y='thalachh', size='chol', sizes=(20, 200), alpha=0.5, data=heart_df)
plt.title('Bubble Graph: Age vs Max Heart Rate (Bubble Size = Cholesterol)')
plt.show()
```

```python
# Box Plot: Age distribution by Chest Pain Type
plt.figure(figsize=(10, 6))
sns.boxplot(x='cp', y='age', data=heart_df , color="red")
plt.title('Box Plot: Age Distribution by Chest Pain Type')
plt.xlabel('Chest Pain Type')
plt.ylabel('Age')
plt.show()
```

```python
#pie chart
plt.pie(x=heart_df["age"][0:10])
plt.show()
```

```python
#swarm plot
sns.swarmplot(x=heart_df["age"])
plt.show()
```
