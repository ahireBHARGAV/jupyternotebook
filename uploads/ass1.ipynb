```python
import pandas as pd
import numpy as np
df = pd.read_csv("dataset_Facebook.csv",sep=";")
```

```python
#lets see data
df.head()
```

```python
#1] subset

subset1= df[["Type","Category"]]
print("Subste 1")
print(subset1.head())

print("Subste 2")
subset2=df[["Paid","comment"]]
print(subset2.head())
```

```python
#2] merge data
#merg_df= pd.concat([subset1, subset2],axis=1)

merg_df=pd.merge(subset1, subset2, left_on="Category", right_on="Paid")
```

```python
merg_df.head()
```

```python
#3] Sort data

pok=merg_df.sort_values(by="comment",ascending=False)
print(pok)
```

```python
#4 trabspose
trans=pok.T
print(trans)
```
